     Introduction
</h2>

The retail sector in India is experiencing rapid growth, driven by shifting consumer behaviors and expanding regional markets. However, managing overall profitability while scaling sales volume presents a significant challenge. This project conducts a comprehensive Exploratory Data Analysis (EDA) on an Indian retail dataset to uncover key sales trends, regional demand variations, and the underlying drivers of operational losses.

<h2 style="background: linear-gradient(90deg, #0F172A, #1E3A8A); color: white; padding: 12px 18px; border-radius: 6px; font-family: sans-serif; margin-top: 0;">
     Aim
</h2>

The primary aim of this project is to perform an end-to-end Exploratory Data Analysis (EDA) to identify business drivers causing systemic losses despite high sales volumes, and to provide data-driven recommendations for improving pricing strategies and regional product distribution.

<h2 style="background: linear-gradient(90deg, #0F172A, #1E3A8A); color: white; padding: 12px 18px; border-radius: 6px; font-family: sans-serif; margin-top: 0;">
    Problem defnition
</h2>

Despite experiencing significant growth in sales volume across multiple categories and regions, the retail business is struggling to achieve expected profitability goals. The business lacks visibility into how individual product pricing, category-level dynamics, and regional factors impact overall business performance. This project aims to analyze transactional data to identify potential revenue leaks, evaluate product-level profitability, and understand regional demand variations to help management make informed strategic decisions

The dataset consists of transaction-level records capturing retail sales operations across India with 25 original attributes:

<h2 style="background: linear-gradient(90deg, #0F172A, #1E3A8A); color: white; padding: 12px 18px; border-radius: 6px; font-family: sans-serif; margin-top: 0;">
    3. Correcting data types
</h2>

```python
#Convert date columns to datetime format
df['Order Date'] = pd.to_datetime(df['Order Date'],format='mixed')
df['Ship Date'] = pd.to_datetime(df['Ship Date'],format='mixed')
```

```python
#Check updated data types
df[['Order Date', 'Ship Date']].dtypes
```

```python
# Convert columns to numeric data type
df['Sales'] = pd.to_numeric(df['Sales'], errors='coerce')
df['List Price'] = pd.to_numeric(df['List Price'], errors='coerce')
df['Quantity'] = pd.to_numeric(df['Quantity'], errors='coerce')
```

<h2 style="background: linear-gradient(90deg, #0F172A, #1E3A8A); color: white; padding: 12px 18px; border-radius: 6px; font-family: sans-serif; margin-top: 0;">
    4. Creating derived columns
</h2>

```python
# Create derived columns
df['Shipping Delay (Days)'] = (df['Ship Date'] - df['Order Date']).dt.days
df['Order Year'] = df['Order Date'].dt.year
df['Order Month'] = df['Order Date'].dt.strftime('%B')
```

```python
# Display new columns to confirm
df[['Order Date', 'Ship Date', 'Shipping Delay (Days)', 'Order Year', 'Order Month']].head()
```

<h2 style="background: linear-gradient(90deg, #0F172A, #1E3A8A); color: white; padding: 12px 18px; border-radius: 6px; font-family: sans-serif; margin-top: 0;">
    5. Outlier detection and handling 
</h2>

```python
sns.boxplot(data=df[['Sales','List Price','Quantity']])
```

```python
Q1 = df['Sales'].quantile(0.25)
Q3 = df['Sales'].quantile(0.75)
IQR = Q3 - Q1 
```

```python
lower_bound = Q1 -1.5*IQR
upper_bound = Q3 +1.5*IQR
outliers =df[(df['Sales']<lower_bound) | (df['Sales']>upper_bound)]
outliers
```

```python
df_clean = df[(df['Sales']>=lower_bound)&(df['Sales']<=upper_bound)]
```

Actionable Business Recommendations
* Audit Profit Margins & Cost Structure: Investigate the underlying costs and discount strategies, as higher sales and list prices are currently yielding lower profitability.

* Focus on High-Margin Products: Re-evaluate unit margins across categories to prioritize products that drive net profit rather than purely chasing high sales volume.

Actionable Business Recommendations

* Audit Discounting Strategies: Stop aggressive promotion campaigns or site-wide discounts on categories like Office Supplies and Clothing that are consistently burning capital across every sales territory.
* Focus Strategy on Profitability over Sales Volume: Shift business goals away from sales revenue alone, as high sales (like Groceries) are currently compounding net financial losses.
